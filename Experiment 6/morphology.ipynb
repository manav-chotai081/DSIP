```python
#in dilation we increase the boundary  (union)
#in erosion we are shrinking or reducing the boundary (intersection of anchor point)
# opening = Erosion + Dilation   (  (A-B) +B )
# closing = Dilation + Erosion   (  (A+B) -B )
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_path = "images.jpg"
input_image = cv2.imread(input_path,0)  #0 for gray scale and 1 for coloured image
plt.figure()
plt.imshow(input_image)
plt.show()
```

```python
input_path = "images.jpg"
image = cv2.imread(input_path,0)  #0 for gray scale and 1 for coloured image

kernel=np.ones((5,5),np.uint8)
```

```python
dilation=cv2.dilate(image,kernel,iterations=3)
plt.subplot(1,2,1)
plt.imshow(dilation,cmap='gray')
```

```python
erosion=cv2.erode(image,kernel,iterations=3)
plt.subplot(1,2,2)
plt.imshow(erosion,cmap='gray')
```

```python
plt.subplot(1,2,1)
plt.imshow(dilation,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(erosion,cmap='gray')
```

```python
opening = cv2.morphologyEx(input_path, cv2.MORPH_OPEN, kernel)

    # Apply Closing
closing = cv2.morphologyEx(input_path, cv2.MORPH_CLOSE, kernel)

plt.subplot(1,2,1)
plt.imshow(opening,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(closing,cmap='gray')
```
